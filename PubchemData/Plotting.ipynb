# This notebook contain all the plots

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## For Viruses

```python
# Directories
dataDir = '/mnt/data.ese/nfs/users/sghosh6/DTRA_project/MACAW/DrugDesignData/'
resultsDir = os.path.join(dataDir, 'Results/virusCommandline_10fold')
```

```python
# Models to plot
models = ['SVR', 'RandomForest', 'XGBoost', 'LightGBM', 'CatBoost', 'NeuralNetwork']

print("Generating parity plots...\n")

for model_name in models:
    # Load data
    df_train = pd.read_csv(os.path.join(resultsDir, f'predictions_train_{model_name}.csv'))
    df_test = pd.read_csv(os.path.join(resultsDir, f'predictions_test_{model_name}.csv'))
    
    # Extract values
    y_train_obs, y_train_pred = df_train['Observed'].values, df_train['Predicted'].values
    y_test_obs, y_test_pred = df_test['Observed'].values, df_test['Predicted'].values
    
    # Calculate metrics
    train_r2 = r2_score(y_train_obs, y_train_pred)
    train_mae = mean_absolute_error(y_train_obs, y_train_pred)
    train_rmse = np.sqrt(mean_squared_error(y_train_obs, y_train_pred))
    test_r2 = r2_score(y_test_obs, y_test_pred)
    test_mae = mean_absolute_error(y_test_obs, y_test_pred)
    test_rmse = np.sqrt(mean_squared_error(y_test_obs, y_test_pred))
    
    # Plot
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.scatter(y_train_obs, y_train_pred, alpha=1, s=20, color='blue', 
               label='Train', edgecolors='blue', linewidth=0.2)
    ax.scatter(y_test_obs, y_test_pred, alpha=1, s=20, color='red', 
               label='Test', edgecolors='red', linewidth=0.2)
    
    # Perfect prediction line
    all_vals = np.concatenate([y_train_obs, y_test_obs, y_train_pred, y_test_pred])
    lim = [all_vals.min(), all_vals.max()]
    ax.plot(lim, lim, 'k--', lw=1, alpha=0.7)
    
    # Formatting
    ax.set_xlabel('pPotency observations', fontsize=10, fontweight='bold')
    ax.set_ylabel('Cross-validated predictions', fontsize=10, fontweight='bold')
    ax.set_title(model_name, fontsize=10, fontweight='bold', pad=10)
    #ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.5)
    #ax.legend(loc='upper center', fontsize=10, frameon = False, framealpha=0.95)

    #legend = ax.legend(loc='upper center', fontsize=10, frameon = False, framealpha=0.95, ncol =2)
    #legend_texts = legend.get_texts()
    #legend_texts[0].set_color('blue')    # Train - blue
    #legend_texts[1].set_color('red')     # Test - red
    
    # Metric boxes
    ax.text(0.02, 0.98, 
            f'$R^2_{{\\mathrm{{Train}}}} = {train_r2:.3f}$\n'
            f'$\\mathrm{{MAE}}_{{\\mathrm{{Train}}}} = {train_mae:.2f}$\n'
            f'$\\mathrm{{RMSE}}_{{\\mathrm{{Train}}}} = {train_rmse:.2f}$',
            transform=ax.transAxes, fontsize=8, va='top', ha='left',
            color='blue',  # Blue text
            #bbox=dict(boxstyle='round,pad=0.7', fc='white', ec='gray', alpha=0.95, lw=1.5)
           )
    
    ax.text(0.98, 0.02, 
            f'$R^2_{{\\mathrm{{Test}}}} = {test_r2:.3f}$\n'
            f'$\\mathrm{{MAE}}_{{\\mathrm{{Test}}}} = {test_mae:.2f}$\n'
            f'$\\mathrm{{RMSE}}_{{\\mathrm{{Test}}}} = {test_rmse:.2f}$',
            transform=ax.transAxes, fontsize=8, va='bottom', ha='right',
            color='red',  # Red text
            #bbox=dict(boxstyle='round,pad=0.7', fc='white', ec='gray', alpha=0.95, lw=1.5)
           )
    
    plt.tight_layout()
    
    # Save
    save_path = os.path.join(resultsDir, f'CL_parity_plot_{model_name}_virus_TrTs')
    plt.savefig(f'{save_path}.png', dpi=300, bbox_inches='tight')
    plt.savefig(f'{save_path}.svg', bbox_inches='tight')
    plt.show()
    plt.close()
    
    print(f" {model_name}")

print(f"\nAll plots saved to: {resultsDir}")
```
