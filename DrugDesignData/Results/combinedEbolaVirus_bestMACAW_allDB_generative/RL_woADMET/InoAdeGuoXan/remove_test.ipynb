```python
import pandas as pd

DF = pd.read_csv("RL_gen2/rl_potency_only_generated_candidates.csv")

cols = [
    "SMILES",
    "pPotency_prediction",
    "seedPotency",
    "deltaPotency",
    "routeLength",
    "reward",
]

print(DF[cols].sort_values("pPotency_prediction", ascending=False).head(20))
print("\nRoute length counts:")
print(DF["routeLength"].value_counts(dropna=False))
print("\nPositive delta count:")
print((DF["deltaPotency"] > 0).sum(), "/", len(DF))
print("Best delta:", DF["deltaPotency"].max())
```
