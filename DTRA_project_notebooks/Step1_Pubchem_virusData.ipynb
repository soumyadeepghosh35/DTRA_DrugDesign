```python
def loadAllCsvsToOneDF(ebolaPubchemDataDir: str | Path) -> pd.DataFrame:
    ebolaPubchemDataDir = Path(ebolaPubchemDataDir)

    DFList = []
    for csvPath in sorted(ebolaPubchemDataDir.glob("*.csv")):
        try:
            DF = pd.read_csv(csvPath, dtype=str, low_memory=False)
            DF.insert(0, "sourceFile", csvPath.name)   # keep provenance
            DFList.append(DF)
        except Exception as e:
            print(f"[SKIP] {csvPath.name}: {e}")

    Ebola_PubchemData = pd.concat(DFList, ignore_index=True, sort=False) if DFList else pd.DataFrame()
    return Ebola_PubchemData


Ebola_PubchemData = loadAllCsvsToOneDF(Ebola_PubchemDataDir)
Ebola_PubchemData['Virus'] = 'Ebola'
```

```python
def loadAllCsvsToOneDF(virusPubchemDataDir: str | Path) -> pd.DataFrame:
    virusPubchemDataDir = Path(virusPubchemDataDir)

    DFList = []
    for csvPath in sorted(virusPubchemDataDir.glob("*.csv")):
        try:
            DF = pd.read_csv(csvPath, dtype=str, low_memory=False)
            DF.insert(0, "sourceFile", csvPath.name)
            DFList.append(DF)
        except Exception as e:
            print(f"[SKIP] {csvPath.name}: {e}")

    return pd.concat(DFList, ignore_index=True, sort=False) if DFList else pd.DataFrame()


# ---- Provide the directory for each virus ----
virusToDirDict = {
    "Ebola":   Ebola_PubchemDataDir,
    "Lassa":   Lassa_PubchemDataDir,
    "Marburg": Marburg_PubchemDataDir,
    "Nipah":   Nipah_PubchemDataDir,
    "SARSCoV": SARSCoV_PubchemDataDir,
    "VEEV":    VEEV_PubchemDataDir,
}

DFList = []
for virusName, virusDir in virusToDirDict.items():
    DF = loadAllCsvsToOneDF(virusDir)
    DF["Virus"] = virusName
    DFList.append(DF)

allVirus_Pubchem = pd.concat(DFList, ignore_index=True, sort=False)
allVirus_Pubchem
```

```python
allVirus_Pubchem = allVirus_Pubchem[['PUBCHEM_CID', 'PUBCHEM_EXT_DATASOURCE_SMILES', 'PubChem Standard Value', 'Standard Type',
                                      'Standard Relation', 'Standard Value', 'Standard Units', 'Virus']]
allVirus_Pubchem = allVirus_Pubchem.dropna(subset=["PUBCHEM_EXT_DATASOURCE_SMILES"])
allVirus_Pubchem
```

```python
# Normalize Standard Units (handles µ, NBSP, spaces, and case)
unitNormalized = (
    allVirus_Pubchem["Standard Units"]
    .astype(str)
    .str.replace("\u00a0", " ", regex=False)   # non-breaking space
    .str.strip()
    .str.replace("µ", "u", regex=False)        # µM -> uM
    .str.replace(" ", "", regex=False)         # remove internal spaces
    .str.lower()
)

# Units that can be converted to molar concentrations for pIC50/pPotency
convertibleUnitsSet = {
    "nm",        # nM
    "um",        # uM (including µM -> uM)
    "mm",        # mM
    "m",         # M
    "umol/l",    # umol/L == uM
    "nmol/l",    # nmol/L == nM (in case it shows up)
    "mmol/l",    # mmol/L == mM (in case it shows up)
    "mol/l",     # mol/L == M (in case it shows up)
}

unitMask = unitNormalized.isin(convertibleUnitsSet)

allVirus_Pubchem = allVirus_Pubchem[unitMask].copy()
allVirus_Pubchem
```

### compute `pPotency = -log10(value_in_M)` to match all the standards

```python
# numeric standard value
standardValueNum = pd.to_numeric(allVirus_Pubchem["Standard Value"], errors="coerce")

# normalize units
unitNormalized = (
    allVirus_Pubchem["Standard Units"]
    .astype(str)
    .str.replace("\u00a0", " ", regex=False)  # NBSP
    .str.strip()
    .str.replace("µ", "u", regex=False)       # µM -> uM
    .str.replace(" ", "", regex=False)
    .str.lower()
)

# unit -> factor to convert to molar
unitToMolarFactorDict = {
    "nm": 1e-9,
    "um": 1e-6,
    "mm": 1e-3,
    "m": 1.0,
    # optional common aliases:
    "umol/l": 1e-6,
    "nmol/l": 1e-9,
    "mmol/l": 1e-3,
    "mol/l": 1.0,
}

molarFactor = unitNormalized.map(unitToMolarFactorDict)

# compute pPotency = -log10(value_in_M)
valueM = standardValueNum * molarFactor
allVirus_Pubchem["pPotency"] = np.where(valueM > 0, -np.log10(valueM), np.nan)
allVirus_Pubchem
```

```python
# Rename columns
allVirus_Pubchem = allVirus_Pubchem.rename(
    columns={'PUBCHEM_EXT_DATASOURCE_SMILES': 'Smiles', 'PUBCHEM_CID': 'compound_id'})
# Reorder columns
allVirus_Pubchem = allVirus_Pubchem[
    ['Virus', 'compound_id', 'Smiles', 'pPotency']]
allVirus_Pubchem
```

### Canonicalize `SMILES`

```python
def canonicalize_smiles(smi):
    mol = Chem.MolFromSmiles(smi)
    return Chem.MolToSmiles(mol, canonical=True) if mol else None

allVirus_Pubchem['Smiles'] = (
    allVirus_Pubchem['Smiles']
    .astype(str)
    .apply(canonicalize_smiles)
)
allVirus_Pubchem
```

### Use pubchem API to fecth data for all target viruses

```python
def requestWithRetry(
    url: str,
    params: Optional[dict] = None,
    timeoutSec: int = 60,
    maxRetries: int = 8,
    baseSleepSec: float = 0.5,
    userAgent: str = "virus-bioassay-harvester/1.0 (email: your_email@domain)"
) -> requests.Response:
    """
    GET with exponential backoff. PubChem can throttle and return HTTP 503. :contentReference[oaicite:3]{index=3}
    """
    headers = {"User-Agent": userAgent}
    lastErr = None

    for attempt in range(maxRetries):
        try:
            resp = requests.get(url, params=params, headers=headers, timeout=timeoutSec)

            if resp.status_code == 503:
                sleepSec = (2 ** attempt) * baseSleepSec + random.random()
                time.sleep(sleepSec)
                continue

            resp.raise_for_status()
            time.sleep(0.2)  # polite pacing
            return resp

        except Exception as e:
            lastErr = e
            sleepSec = (2 ** attempt) * baseSleepSec + random.random()
            time.sleep(sleepSec)

    raise RuntimeError(f"Request failed after retries. url={url} lastErr={lastErr}")


def esearchPcAssayAids(
    term: str,
    retmax: int = 200,
    toolName: str = "virusBioassayHarvester",
    email: str = "your_email@domain",
    apiKey: Optional[str] = None
) -> List[int]:
    """
    Find PubChem BioAssay IDs (AIDs) via NCBI ESearch:
    Base: https://eutils.ncbi.nlm.nih.gov/entrez/eutils/ :contentReference[oaicite:4]{index=4}
    Use tool/email (and optionally api_key) per E-utilities guidelines. :contentReference[oaicite:5]{index=5}
    """
    url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi"
    params = {
        "db": "pcassay",
        "term": term,
        "retmax": retmax,
        "retmode": "json",
        "tool": toolName,
        "email": email,
    }
    if apiKey:
        params["api_key"] = apiKey

    resp = requestWithRetry(url, params=params)
    data = resp.json()
    idList = data.get("esearchresult", {}).get("idlist", [])
    return [int(x) for x in idList]


def fetchConciseAssayCsv(aid: int) -> pd.DataFrame:
    """
    PubChem PUG-REST concise table for one assay:
    https://pubchem.ncbi.nlm.nih.gov/rest/pug/assay/aid/<AID>/concise/CSV :contentReference[oaicite:6]{index=6}
    """
    url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/assay/aid/{aid}/concise/CSV"
    resp = requestWithRetry(url)
    DF = pd.read_csv(io.StringIO(resp.text))
    DF["AID"] = aid
    return DF


def buildAllVirusPubchem(
    virusQueryMap: Dict[str, str],
    aidsPerVirus: int = 200,
    onlyActive: bool = False,
    toolName: str = "virusBioassayHarvester",
    email: str = "your_email@domain",
    apiKey: Optional[str] = None
) -> pd.DataFrame:
    rows = []

    for virusName, entrezTerm in virusQueryMap.items():
        aids = esearchPcAssayAids(
            term=entrezTerm,
            retmax=aidsPerVirus,
            toolName=toolName,
            email=email,
            apiKey=apiKey
        )

        for aid in aids:
            try:
                DF = fetchConciseAssayCsv(aid)
                DF["Virus"] = virusName
                DF["EntrezTerm"] = entrezTerm
                rows.append(DF)
            except Exception:
                # Skip failures; they are often caused by very large or unusual assays
                continue

    if not rows:
        return pd.DataFrame()

    DF = pd.concat(rows, ignore_index=True)

    # Optional filter: keep only rows with an "Active" outcome if the column exists
    if onlyActive:
        outcomeColCandidates = [c for c in DF.columns if c.strip().lower() in ["activityoutcome", "outcome"]]
        if outcomeColCandidates:
            outcomeCol = outcomeColCandidates[0]
            DF = DF[DF[outcomeCol].astype(str).str.lower().eq("active")].copy()

    # User-requested name
    allVirus_pubchem = DF
    return allVirus_pubchem


# ---------------------------
# Customize your virus terms
# ---------------------------
virusQueryMap = {
    "Ebola": '(Ebola OR Ebolavirus OR "Zaire ebolavirus") AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
    "Marburg": '(Marburg OR Marburgvirus) AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
    "Lassa": '(Lassa OR "Lassa virus") AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
    "VEEV": '("Venezuelan equine encephalitis" OR VEEV) AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
    "Nipah": '(Nipah OR "Nipah virus") AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
    "Hanta": '(Hantavirus OR "Hanta virus") AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
    "SARS-CoV": '("SARS-CoV" OR "SARS coronavirus") AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
    "SARS-CoV-2": '("SARS-CoV-2" OR COVID-19 OR "2019-nCoV") AND (antiviral OR inhibitor OR neutralization OR IC50 OR EC50)',
}

# Build the dataset
allVirus_pubchem = buildAllVirusPubchem(
    virusQueryMap=virusQueryMap,
    aidsPerVirus=150,   # increase carefully; SARS-CoV-2 can explode in size
    onlyActive=False,   # set True if you want only active outcomes (when available)
    toolName="virusBioassayHarvester",
    email="sghosh6@lbl.gov",
    apiKey="0ffd39fe99552944ce9a9204efef27be7109"         # add an NCBI api_key if you will run high-throughput calls :contentReference[oaicite:7]{index=7}
)

# Optional: persist
# allVirus_pubchem.to_parquet("allVirus_pubchem.parquet", index=False)
# allVirus_pubchem.to_csv("allVirus_pubchem.csv", index=False)

allVirus_pubchem
```

### Add `SMILES` from Pubchem to all the BioAssay data

```python
def requestWithRetry(
    url: str,
    params: Optional[dict] = None,
    timeoutSec: int = 60,
    maxRetries: int = 10,
    baseSleepSec: float = 0.6,
    userAgent: str = "pubchem-smiles-fetcher/1.0 (email: your_email@domain)",
) -> requests.Response:
    headers = {"User-Agent": userAgent}

    for attempt in range(maxRetries):
        try:
            resp = requests.get(url, params=params, headers=headers, timeout=timeoutSec)

            # PubChem may throttle; backoff helps. :contentReference[oaicite:2]{index=2}
            if resp.status_code == 503:
                time.sleep((2**attempt) * baseSleepSec + random.random())
                continue

            resp.raise_for_status()
            time.sleep(0.15)  # polite pacing
            return resp

        except Exception:
            time.sleep((2**attempt) * baseSleepSec + random.random())

    raise RuntimeError(f"Failed after {maxRetries} retries: {url}")


def chunked(items: List[int], chunkSize: int) -> List[List[int]]:
    return [items[i : i + chunkSize] for i in range(0, len(items), chunkSize)]


def fetchSmilesForCidChunk(cidChunk: List[int], debug: bool = False) -> pd.DataFrame:
    """
    Fetch SMILES-like properties for multiple CIDs.
    Note: PubChem may return 'ConnectivitySMILES' instead of 'CanonicalSMILES'. :contentReference[oaicite:3]{index=3}
    """
    cidStr = ",".join(str(x) for x in cidChunk)

    # Request both Isomeric and Connectivity SMILES explicitly
    url = (
        "https://pubchem.ncbi.nlm.nih.gov/rest/pug/"
        f"compound/cid/{cidStr}/property/IsomericSMILES,ConnectivitySMILES,CanonicalSMILES/CSV"
    )

    resp = requestWithRetry(url)
    text = resp.text.lstrip()

    if not (text.startswith("CID,") or text.startswith('"CID"')):
        raise ValueError(f"Non-CSV response (first 200 chars): {text[:200]}")

    DF = pd.read_csv(io.StringIO(resp.text))

    if debug:
        print("[DEBUG] Returned columns:", DF.columns.tolist())
        print(DF.head(5))

    # Prefer isomeric (stereo), fall back to connectivity/canonical
    smilesColCandidates = [
        "IsomericSMILES",
        "ConnectivitySMILES",  # commonly returned for "canonical-like" SMILES :contentReference[oaicite:4]{index=4}
        "CanonicalSMILES",
        "SMILES",
    ]
    DF["smiles"] = pd.NA
    for col in smilesColCandidates:
        if col in DF.columns:
            DF["smiles"] = DF["smiles"].fillna(DF[col])

    return DF[["CID", "smiles"]]


def sanityCheckSmilesFetch(cidList: List[int], nTest: int = 10) -> None:
    # Known-good connectivity sanity: Aspirin CID 2244 returns ConnectivitySMILES. :contentReference[oaicite:5]{index=5}
    knownUrl = "https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/2244/property/CanonicalSMILES/CSV"
    knownResp = requestWithRetry(knownUrl)
    print("[SANITY] Known CID (2244) response head:", knownResp.text[:120])

    testCids = cidList[:nTest]
    DF = fetchSmilesForCidChunk(testCids, debug=True)

    gotCount = int(DF["smiles"].notna().sum())
    print(f"[SANITY] Requested {len(testCids)} CIDs, received {len(DF)} rows, SMILES present for {gotCount} rows.")
    print(DF)

    if gotCount == 0:
        raise RuntimeError(
            "Sanity check failed: no SMILES columns were populated. "
            "The debug print above should show which SMILES column name PubChem returned."
        )


def fetchSmilesMapForCids(
    cidList: List[int],
    chunkSize: int = 50,
    verbose: bool = True,
) -> Dict[int, str]:
    smilesMap: Dict[int, str] = {}

    for cidChunk in chunked(cidList, chunkSize):
        try:
            DF = fetchSmilesForCidChunk(cidChunk, debug=False)
            for cid, smi in zip(DF["CID"], DF["smiles"]):
                if pd.notna(cid) and pd.notna(smi):
                    smilesMap[int(cid)] = str(smi)
        except Exception as e:
            if verbose:
                print(f"[WARN] Chunk failed (n={len(cidChunk)}): {e}")

    return smilesMap


# ---------------------------
# Main: sanity check -> full run
# ---------------------------
DF = allVirus_pubchem.copy()

DF["CID"] = pd.to_numeric(DF["CID"], errors="coerce")
uniqueCids = (
    DF["CID"]
    .dropna()
    .astype(int)
    .loc[lambda s: s > 0]
    .drop_duplicates()
    .tolist()
)

# 1) Sanity check (first 10)
sanityCheckSmilesFetch(uniqueCids, nTest=10)

# 2) Full fetch
smilesMap = fetchSmilesMapForCids(uniqueCids, chunkSize=50, verbose=True)

# 3) Add SMILES
DF["smiles"] = DF["CID"].astype("Int64").map(smilesMap)
allVirus_pubchem = DF

print(allVirus_pubchem[["CID", "smiles"]].head(10))
print("Missing SMILES rows:", int(allVirus_pubchem["smiles"].isna().sum()))
```
