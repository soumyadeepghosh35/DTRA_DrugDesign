### Marge the data for `DORAnet` generated molecules

```python
%%time

# Concatenate in chunks and free memory ---
DF_list = [
    basidalinPrecursor_DORAnetGeneratedDF,
    InosineSimilarstructures_DORAnetGeneratedDF,
    InosineSubstructures_DORAnetGeneratedDF,
    drugBank_DORAnetGeneratedDF,
    drugCentral_DORAnetGeneratedDF,
]

allDORAnetGeneratedMolecules = pd.concat(DF_list, ignore_index=True)

# Free memory from individual dataframes
del basidalinPrecursor_DORAnetGeneratedDF
del InosineSimilarstructures_DORAnetGeneratedDF
del InosineSubstructures_DORAnetGeneratedDF
del drugBank_DORAnetGeneratedDF
del drugCentral_DORAnetGeneratedDF
gc.collect()

print(f"Total molecules before deduplication: {len(allDORAnetGeneratedMolecules)}")

# --- 2. Only keep needed columns early ---
#needed_cols = ['SMILES', 'Is_Starter', 'SourceDirectories', 'source_DataBase']
#allDORAnetGeneratedMolecules = allDORAnetGeneratedMolecules[needed_cols]

# --- 3. Batch canonicalize SMILES using vectorized approach ---
def canonicalize_smiles_batch(smiles_series, batch_size=50000):
    results = []
    for i in range(0, len(smiles_series), batch_size):
        batch = smiles_series.iloc[i:i + batch_size]
        canonical = []
        for smi in batch:
            try:
                mol = Chem.MolFromSmiles(smi)
                if mol is not None:
                    canonical.append(Chem.MolToSmiles(mol))
                else:
                    canonical.append(None)
            except:
                canonical.append(None)
        results.extend(canonical)
        print(f"  Processed {min(i + batch_size, len(smiles_series))}/{len(smiles_series)} SMILES...")
        gc.collect()
    return results

print("Canonicalizing SMILES...")
allDORAnetGeneratedMolecules['SMILES'] = canonicalize_smiles_batch(allDORAnetGeneratedMolecules['SMILES'])

# --- 4. Drop invalid and duplicate SMILES ---
allDORAnetGeneratedMolecules = allDORAnetGeneratedMolecules.dropna(subset=['SMILES'])
allDORAnetGeneratedMolecules = allDORAnetGeneratedMolecules.drop_duplicates(subset='SMILES', keep='first')
allDORAnetGeneratedMolecules = allDORAnetGeneratedMolecules.reset_index(drop=True)
gc.collect()

print(f"Total molecules after deduplication: {len(allDORAnetGeneratedMolecules)}")

# --- 5. Fix: save to a file path string, not the dataframe variable ---
output_path = DORANETmoleculesDataDir + "allDORAnetGeneratedMolecules.csv"
allDORAnetGeneratedMolecules.to_csv(output_path, index=False, encoding="utf-8")
print(f"Saved to {output_path}")

allDORAnetGeneratedMolecules
```
